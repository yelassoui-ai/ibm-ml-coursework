## Import the required libraries


````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import fetch_california_housing
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error, root_mean_squared_error, mean_absolute_error, r2_score
from scipy.stats import skew
````

### Load the California Housing data set


````python
# Load the dataset
data = fetch_california_housing()
X, y = data.data, data.target
````

````python
print(data.DESCR)
````

Reserve 20% of the data for evaluation

### How are the median house prices distributed?


````python
# Plot the distribution
plt.hist(1e5*y_train, bins=30, color='lightblue', edgecolor='black')
plt.title(f'Median House Value Distribution\nSkewness: {skew(y_train):.2f}')
plt.xlabel('Median House Value')
plt.ylabel('Frequency')
````

Evidently the distribution is skewed and there are quite a few clipped values at around $500,000. 


### Model fitting and prediction
Let's fit a random forest regression model to the data and use it to make median house price predicions.
Use the default parameters, which includes using 100 base estimators, or regression trees.


````python
# Initialize and fit the Random Forest Regressor
rf_regressor = RandomForestRegressor(n_estimators=100, random_state=42)
rf_regressor.fit(X_train, y_train)

# Predict on test set
y_pred_test = rf_regressor.predict(X_test)
````

### Estimate out-of-sample MAE, MSE, RMSE, and R²


````python
mae = mean_absolute_error(y_test, y_pred_test)
mse = mean_squared_error(y_test, y_pred_test)
rmse = root_mean_squared_error(y_test, y_pred_test)
r2 = r2_score(y_test, y_pred_test)
print(f"Mean Absolute Error (MAE): {mae:.4f}")
print(f"Mean Squared Error (MSE): {mse:.4f}")
print(f"Root Mean Squared Error (RMSE): {rmse:.4f}")
print(f"R² Score: {r2:.4f}")
````

### Exercise 3. What do these statistics mean to you?

- **RMSE = 0.5053 → about \$50,530.** RMSE is larger than MAE because it penalizes large errors more, which suggests that some predictions are off by a lot.

### Exercise 6. What trend can you infer from this residual plot?


The overall average residual is small (about -\$1,200), but the plot shows a clear trend: **the residuals increase from negative to positive as the actual median house value increases.**

- **Low-priced areas are overpredicted** (negative residuals). For houses under \$100k, the average residual is about -\$20k.
- **High-priced areas are underpredicted** (positive residuals). For houses above \$400k, the average residual is about +\$69k.
- The effect is strongest at the **\$500k cap**: the model can't predict the clipped values, so it underestimates them.

So the model is biased toward the middle of the price range. The small average error hides this, because the over- and underpredictions cancel out.

### Exercise 7. Display the feature importances as a bar chart.
Do you think these feature weights have practial significance? Are any of the features possibly sharing importance with other correlated features?


````python
# Enter your code here 

# Feature importances
importances = rf_regressor.feature_importances_
indices = np.argsort(importances)[::-1]
features = data.feature_names

# Plot feature importances
plt.bar(range(X.shape[1]), importances[indices],  align="center")
plt.xticks(range(X.shape[1]), [features[i] for i in indices], rotation=45)
plt.xlabel("Feature")
plt.ylabel("Importance")
plt.title("Feature Importances in Random Forest Regression")
plt.show()
````

- **MedInc (about 0.52)** is by far the most important feature. That makes practical sense, because richer neighbourhoods have more expensive houses.
- **AveOccup (about 0.14)** comes second.
- **Latitude and Longitude (about 0.09 each)** have almost equal importance. Together they describe one underlying feature, **location**, so their importance is split between them. Combined (about 0.18), location would actually be the **second most important** factor, above AveOccup. Replacing lat/lng with a categorical location (city, suburb, etc.) would probably show this more clearly.
- **AveRooms and AveBedrms** are probably strongly correlated with each other, so they likely share importance too, and each looks less important than it really is.

So the weights are useful as a rough guide, but correlated features split importance between them. A correlation matrix would be needed before drawing firm conclusions.

### Exercise 8. Some final thoughts to consider

- Will the skewness affect the performance of Random Forest regression?
- Does the clipping of median house prices above \$500,000 bias the predictions?
- Also, do we need to standardize the data?


- **Skewness:** Random forest regression is quite robust to skewed distributions and outliers. Unlike linear regression, it makes no assumption about how the data is distributed, because it just splits the data into regions. The skewed target does mean there are fewer training examples at high prices, which adds to the underprediction of expensive houses seen in the residual plot.
- **Clipping at \$500,000:** Yes, this biases the predictions. The clipped values have no variability: many very different expensive areas all get the same label, \$500k. Because each prediction is an average of tree outputs, the model can never predict above the cap, so these houses are systematically underpredicted. The clipped values also distort the evaluation metrics. Removing them during preprocessing (or modelling them separately) would let the model explain the real variance better.
- **Standardization:** Not needed. Tree splits only depend on the order of the values within each feature, not their scale. Standardization matters for distance-based models such as KNN or SVMs, and for regularized linear models.

<details><summary>Click here for the solution</summary>
Compared to linear regression, random forest regression is quite robust against outliers and skewed distributions.
This is because random forest regression doesn't make any assumptions about the data distribution, where linear regression performs best with normally distributed data.

Standardizing the data isn't necessary like it is for distance-based algortihms like KNN or SVMs.

Regarding the clipped vlaues, there is no variablilty in those values. Removing them in preprocessing might help the model to better explain the actual variance in the data. 

The clipped values can also bias the predictions.
Also these clipped values can mislead evaluation metrics. As you've learned from this lab, it's crucially important for you to visualize your results.
